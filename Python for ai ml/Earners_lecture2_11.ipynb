#Map

```python
# map(function, iterable) --> it returns a map object --> convert it into list --> list(map(...))
```

```python
# write a function to return a list of the cube of every number in the given list .
```

```python
def cube(Ist):
  cb=[]
  for elem in Ist:
    cb.append(elem**3)
  return cb
```

```python
cube([5,7,3])
```

```python
def cb(n):
  return n**3
cube=[]
Ist=[4,6,2]
for elem in Ist:
  cube.append(cb(elem))
print(cube)
```

```python
cube=list(map(lambda x: x**3,[4,6,2]))
```

#Filter

```python
num =[1,2,3,4,5,6,7]
# write a code to create a list of all the even numbers of the given list
even=[]
for n in num:
  if n%2 ==0:
    even.append(n)
print(even)
```

```python
num =[1,2,3,4,5,6,7]
even =list(filter(lambda x:x%2 == 0,num ))
```

```python
print(even)
```

```python
even=list(filter(lambda x: x%2 ==0 , [1,2,3,4,5,6,7]))
print(even)
```
