```python
# Given a string ,write a code to convert it to lowercase , u cannot use lower()
```

```python
s = "Raj"
new_s = ""

for ch in s:
    if "A" <= ch <= "Z":  # Check if character is uppercase
        new_s += chr(ord(ch) + 32)  # Convert ASCII value to lowercase
    else:
        new_s += ch

print(new_s)
```
