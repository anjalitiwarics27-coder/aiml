```python
def calc_grade(marks):
  if marks>90:
    grade ="A"
  elif marks>75:
    grade ="B"
  elif marks>60:
    grade ="C"
  else:
    grade ="D"
  return grade
```

```python
class Student:
  def __init__(self,roll,name,age,course,marks):
    self.roll=roll
    self.name=name
    self.age=age
    self.course=course
    self.marks=marks
  def show_student(self):
    print(self.roll,self.name,self.age,self.course,self.marks)
```

```python
# Build Complete Student Management Menu
# 2 Display Students
# 3 Search Student
# 4 Find Topper
# 5 Calculate Average
# 6 Exit
```

```python
def search(roll_no):
  for student in students:
    if student["roll"]==roll_no:
      print(student)
      return "Record Found"
  return "Not Found"
def find_topper(students):
  mx_marks=0
  topper={}
  for student in students:
    if student["marks"]>mx_marks:
      topper=student
  return topper
```

```python
while True:
  choice=int(input("Enter your choice: "))
  if choice==1:
    print(students)
  elif choice==2:
    roll=int(input("Enter Roll No: "))
    ans=search(roll)
    print(ans)
  else:
    print("Invalid Choice")
    break
```

```python
while True:
  choice = int(input("Enter your choice: "))
  if choice == 2:
    for student in students:
      print(student)
  elif choice == 3:
    roll = int(input("Enter student Roll no: "))
    print(search(roll))
  elif choice == 4:
    print(find_topper(students))
  elif choice == 5:
    total = 0
    for student in students:
      total += student["marks"]
    print("Average:", total / len(students))
  elif choice == 6:
    print("Thank You!")
  else:
    print("Invalid Choice")
    break
```
