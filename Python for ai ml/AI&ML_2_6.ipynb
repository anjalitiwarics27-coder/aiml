#Problem 9: Most Frequent Character

```python
# Problem Statement
# Find the character that appears the maximum number of times in a string.
# Ignore spaces
```

```python
s = "Python is the most amazing programming language for ML"
freq = {}
for c in s:
  if c != " ":
    if c not in freq:
      freq[c] = 1
    else:
      freq[c] += 1
print(freq)
```

```python
# __init__()
# __init__() --> is built-in method which is always executed when the class gets initiated (object is being getting createed)
```

```python
class Person:
  def __init__(self, name, city):
    self.name = name
    self.city = city
```

```python
p1=Person("Raj","Mumbai")
print(p1.name,p1.city)
```

```python
p2 = Person("Vivek", "Pune")
print(p2.name, p2.city)
```

```python
p2 = Person("Riya", "Delhi")
print(p2.name, p2.city)
```
