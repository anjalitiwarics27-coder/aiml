```python
marks=[]
for i in range(5):
  m=int(input())
  marks.append(m)
```

```python
s=sum(marks)
n=len(marks)
avg=s/n
print(avg)
```

```python
mn=min(marks)
mx=max(marks)
print(mn,mx)
```

```python

# | Average   | Grade |
# | --------  | ----- |
# | 90+       | A     |
# | 75-89     | B     |
# | 60-74     | C     |
# | Below 60  | D     |
```

```python
if avg>=90:
  grade="A"
elif avg>=75:
  grade="B"
else:
  grade="D"
print(grade)
```

```python
#Final systamtic code:
marks = []

for i in range(5):
    mark = int(input(f"Enter mark {i+1}: "))
    marks.append(mark)  # Fixed: append 'mark' (the variable)

average = sum(marks) / len(marks)

print("Average :", average)
print("Highest :", max(marks))
print("Lowest  :", min(marks))

if average >= 90:
    print("Grade   : A")
elif average >= 75:
    print("Grade   : B")
elif average >= 60:
    print("Grade   : C")
else:
    print("Grade   : D")
```

# Problem 2 :Word Frequency Counter

```python
# Problem statement
# Take s sentence as input.
# Display how many times each word appears.
# Eg1: python is fun python is easy
# op: python- 1,is -2 , easy -1 , fun -1
```

```python
s=input()
words=s.split(" ")
print(words)
```

```python
freq={}
for word in words:
  if word in freq:
    freq[word]+=1
  else:
    freq[word]=1
print(freq)
```

#Problem 4 :Character Counter

```python
# Problem statement
# TAKE A STRING AS INPUT FROM THE USER:
# display
# Number of vowels
# Number of consonants
# Number of digits
# Number of spaces
```

```python
# Tp1: Python3 is Fun
# Op1: vowels-3
# consonants-8
# digits-1
# space-2
```

```python
from string import digits
s=input()
s1=s.lower()
vowels=0
consonants=0
digits=0
space=0
for c in s1:
  if c in "aeiou":
    vowels+=1
  elif c>="a"and c<="z":
    consonants+=1
  elif c>="0" and c<="9":
    digits+=1
  elif c==" ":
    space+=1
print("Vowels - ",vowels)
print("Consonants -",consonants)
print("Digits -",digits)
print("Space -",space)
```

```python
text=input()
vowels=0
consonants=0
digits=0
space=0
for ch in text:
  if ch.lower() in"aeiou":
    vowels+=1
  elif ch.isalpha():
    consonants+=1
  elif ch.isdigit():
    digits+=1
  elif ch.isspace():
    space+=1
print("Vowels - ",vowels)
print("Consonants -",consonants)
print("Digits -",digits)
print("Space -",space)
```

#Problem 5 : Shopping Bill

```python
# Take item names and prices for 5 products.
# Storeitem names in one list
# prices in another list
# display
# total bill amount
# highest proced item
# lowest priced item
```

```python
# Eg1:
# Apple
# 100
# Milk
# 50
# Bread
# 40
# Rice
# 80
# Juice
# 120
# Op1: 390
# Juice
# Bread
```

```python
names=[]
prices=[]
for i in range(5):
  name=input()
price=int(input())
names.append(name)
prices.append(price)
print(names)
print(prices)
```

```python
total_bill_amount = sum(prices)

# Find max and min based on prices
max_price, max_prd = max(zip(prices, names))
min_price, min_prd = min(zip(prices, names))

print("Total Bill Amount:", total_bill_amount)
print("Highest Priced Item:", max_price, max_prd)
print("Lowest Priced Item :", min_price, min_prd)
```

```python
total_bill_amount = sum(prices)

mx = prices[0]
mx_prd = names[0]  # Initialize with the first product name
mn = prices[0]
mn_prd = names[0]  # Initialize with the first product name

for i in range(len(prices)):  # Use len(prices) instead of hardcoded 5
    if prices[i] > mx:
        mx_prd = names[i]
        mx = prices[i]
    if prices[i] < mn:
        mn_prd = names[i]
        mn = prices[i]

print("Total Bill:", total_bill_amount)
print("Highest:", mx, mx_prd)
print("Lowest :", mn, mn_prd)
```
