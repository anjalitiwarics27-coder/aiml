```python
# A college stores marks of 4 students in 3 subjects: Maths, Science, and English.
# import numpy as np
# marks = np.array([
# [72, 85, 90],
# [65, 78, 82],
# [88, 91, 84],
# [70, 76, 80]
# ])
# Perform the following:
# Display the marks of the third student.
# Display the Science marks of the second student.
# Add 5 bonus marks to every student's English marks.
# Change the Maths marks of the fourth student to 75.
# Display the final marks array.
```

```python
import numpy as np
marks = np.array([
    [72, 85, 90],
[65, 78, 82],
[88, 91, 84],
[70, 76, 80]
])
print(marks[2])
print(marks[1,1])
marks[:,2]+=5
print(marks)
marks[3,0]=75
print(marks)
```

```python
import numpy as np

marks = np.array([
    [72, 85, 90],
    [65, 78, 82],
    [88, 91, 84],
    [70, 76, 80]
])

# 1. Display marks of the 3rd student (index 2)
print("3rd Student Marks:", marks[2])

# 2. Display Science marks (column index 1) of the 2nd student (row index 1)
print("2nd Student Science Marks:", marks[1, 1])

# 3. Add 5 bonus marks to every student's English marks (column index 2)
marks[:, 2] += 5
print(marks)
# 4. Change Maths marks (column index 0) of the 4th student (row index 3) to 75
marks[3, 0] = 75

# 5. Display final array
print("\nFinal Marks Array:")
print(marks)
```
