```python
# A simple grayscale image can be represented as a 2D NumPy array where each number represents pixel intensity.
# image = np.array([
# [10, 20, 30, 40, 50],
# [20, 30, 40, 50, 60],
# [30, 40, 50, 60, 70],
# [40, 50, 60, 70, 80],
# [50, 60, 70, 80, 90]
# ])
# Tasks
# Find the dimensions of the image.
# Find its shape.
# Find the total number of pixels.
# Display the center pixel.
# Extract the top-left 3 × 3 region.
# Extract the bottom-right 2 × 2 region.
# Increase the intensity of the center 3 × 3 region by 20.
# Display the modified image.
```

```python
import numpy as np

# Define the image array
image = np.array([
    [10, 20, 30, 40, 50],
    [20, 30, 40, 50, 60],
    [30, 40, 50, 60, 70],
    [40, 50, 60, 70, 80],
    [50, 60, 70, 80, 90]
])

# 1. Dimensions of the image
dimensions = image.ndim
print("Dimensions:", dimensions)

# 2. Shape of the image
shape = image.shape
print("Shape:", shape)

# 3. Total number of pixels
total_pixels = image.size
print("Total number of pixels:", total_pixels)

# 4. Center pixel
center_pixel = image[2, 2]
print("Center pixel:", center_pixel)

# 5. Extract top-left 3 x 3 region
top_left_3x3 = image[:3, :3]
print("Top-left 3x3 region:\n", top_left_3x3)

# 6. Extract bottom-right 2 x 2 region
bottom_right_2x2 = image[-2:, -2:]
print("Bottom-right 2x2 region:\n", bottom_right_2x2)

# 7. Increase intensity of center 3 x 3 region by 20
image[1:4, 1:4] += 20

# 8. Display modified image
print("Modified image:\n", image)
```

```python
# A company stores quarterly revenue and expenses:
# financials = np.array([
# [120, 80],
# [150, 90],
# [180, 110],
# [200, 130]
# ])
# Rows = Q1, Q2, Q3, Q4
# Columns = Revenue, Expense
# Tasks
# Display Q2's complete financial data.
# Display all quarterly revenues.
# Display all quarterly expenses.
# Extract Q2 and Q3 data.
# Increase Q4 revenue by 15%.
# Reduce Q1 expense by 10%.
# Display the final financial matrix.
```

```python
import numpy as np

# Initial financial matrix
financials = np.array([
    [120, 80],   # Q1
    [150, 90],   # Q2
    [180, 110],  # Q3
    [200, 130]   # Q4
])

# 1. Display Q2's complete financial data
print("Q2 Data:", financials[1])

# 2. Display all quarterly revenues (Column 0)
print("Revenues:", financials[:, 0])

# 3. Display all quarterly expenses (Column 1)
print("Expenses:", financials[:, 1])

# 4. Extract Q2 and Q3 data
print("Q2 and Q3 Data:\n", financials[1:3])

# 5. Increase Q4 revenue by 15%
financials[3, 0] = financials[3, 0] * 1.15

# 6. Reduce Q1 expense by 10%
financials[0, 1] = financials[0, 1] * 0.90

# 7. Display the final financial matrix
print("\nFinal Financial Matrix:\n", financials)
```

```python
# A classroom has 6 rows and 5 seats per row.
# Initially, all seats are empty.
# 0 = Empty
# 1 = Occupied
# Tasks
# Create a 6 × 5 array filled with zeros.
# Mark the first seat of the first row as occupied.
# Mark all seats in the 3rd row as occupied.
# Mark the last two seats of the last row as occupied.
# Display the first 3 rows.
# Display the last 2 columns.
# Display the complete seating matrix .
```

```python
import numpy as np

# 1. Create a 6 x 5 array filled with zeros
seating = np.zeros((6, 5), dtype=int)

# 2. Mark the first seat of the first row as occupied
seating[0, 0] = 1

# 3. Mark all seats in the 3rd row as occupied (row index 2)
seating[2, :] = 1

# 4. Mark the last two seats of the last row as occupied
seating[-1, -2:] = 1

# 5. Display the first 3 rows
print("First 3 rows:\n", seating[:3])

# 6. Display the last 2 columns
print("\nLast 2 columns:\n", seating[:, -2:])

# 7. Display the complete seating matrix
print("\nComplete Seating Matrix:\n", seating)
```
